## Data Preprocessing

####Date Formatting

```python
for df in [train, features, test]:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
for name, df in zip(["train", "features", "test"], [train, features, test]):
    print(f"{name} Date dtype:", df["Date"].dtype)
    print(f"{name} Date nulls after conversion:", df["Date"].isna().sum())
    print()
```

####Merge Datasets

Train Set

```python
train_merged = train.merge(features, on=["Store", "Date"], how="left")
train_merged = train_merged.merge(stores, on="Store", how="left")
```

```python
train_merged.shape
```

```python
train_merged.head(5)
```

Test Test

```python
test_merged = test.merge(features, on=["Store", "Date"], how="left")
test_merged = test_merged.merge(stores, on="Store", how="left")
```

```python
test_merged.shape
```

```python
test_merged.head(5)
```

###Handling Missing Values

Train Set

```python
train_merged.isnull().sum()
```

```python
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']
```

```python
train_merged[markdown_cols] = train_merged[markdown_cols].fillna(0)
```

```python
train_merged.isnull().sum()
```

```python
train_merged.info()
```

Test Set

```python
test_merged.isnull().sum()
```

Handling Markdowns

```python
test_merged[markdown_cols] = train_merged[markdown_cols].fillna(0)
```

Handling CPI and Unemployment

```python
# Sort test set
test_merged = test_merged.sort_values(['Store','Dept','Date'])

# Interpolate per Store-Dept combination
test_merged['CPI'] = test_merged.groupby(['Store','Dept'])['CPI'].transform(lambda x: x.interpolate(method='linear'))
test_merged['Unemployment'] = test_merged.groupby(['Store','Dept'])['Unemployment'].transform(lambda x: x.interpolate(method='linear'))

# Forward/backward fill remaining NaNs
test_merged['CPI'] = test_merged['CPI'].ffill().bfill()
test_merged['Unemployment'] = test_merged['Unemployment'].ffill().bfill()

# Reset index
test_merged.reset_index(drop=True, inplace=True)
```

```python
test_merged.isnull().sum()
```

## Encoding Categorial Data

Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
train_merged['Type'] = encoder.fit_transform(train_merged['Type'])
test_merged['Type'] = encoder.transform(test_merged['Type'])
```

Boolean Encoding

```python
bool_cols = ['IsHoliday_x', 'IsHoliday_y']
for col in bool_cols:
    train_merged[col] = train_merged[col].astype(int)
    test_merged[col] = test_merged[col].astype(int)
```
